# Tema 01 — Sintaxis: Tu primer programa en Python

### 🧠 Conceptos clave

| Concepto | Definición |
|---|---|
| **Sintaxis** | Las reglas gramaticales de Python. Si las violas, el código no se ejecuta. |
| **Instrucción** | Una línea de código que le dice a Python que haga algo. |
| **`print()`** | Función que muestra un mensaje en la pantalla (la consola). |
| **Error de sintaxis** | Mensaje que Python lanza cuando algo está mal escrito. |

### 📖 Explicación

Imagina que Python es un funcionario muy estricto que lee tus instrucciones de arriba hacia abajo, una por una. Si encuentra una frase mal escrita, se detiene inmediatamente y te dice **dónde** está el problema. No adivina lo que quisiste decir: necesita que le hables con precisión.

Tu primera herramienta es `print()`: una función que le dice a Python "muestra este mensaje en la pantalla". Todo lo que quieras mostrar va **dentro de los paréntesis** y, si es texto, **entre comillas**.

````python
# Tu primer programa en Python
# Las líneas que empiezan con # son comentarios: Python las ignora
# Son como notas al margen en un documento

print("¡Hola, Ciencias Políticas!")
print("Bienvenidos a la Universidad de Antioquia")
print("Este es mi primer programa en Python")
````

````python
# Python ejecuta línea por línea, de arriba hacia abajo
print("Línea 1: Primero se ejecuta esto")
print("Línea 2: Luego esto")
print("Línea 3: Y por último esto")

# El ORDEN importa: si cambias las líneas, cambia el resultado
````

````python
# print() puede mostrar números directamente, sin comillas
print(2025)
print(3.14)

# Y también puede hacer operaciones dentro de los paréntesis
print(10 + 5)
print("Resultado de 10 + 5:", 10 + 5)  # Texto y cálculo juntos
````

# Tema 02 — Variables: Asignación, Nomenclatura y Convenciones

### 🧠 Conceptos clave

| Concepto | Definición |
|---|---|
| **Variable** | Espacio en memoria donde se almacena un valor. Piensa en ella como una *caja etiquetada*. |
| **Operador `=`** | Asigna el valor de la derecha a la variable de la izquierda. No significa "igual a" en matemáticas. |
| **Case Sensitive** | Python distingue mayúsculas de minúsculas: `Votos` ≠ `votos`. |
| **Sobrescritura** | Asignar un nuevo valor a una variable existente reemplaza el dato anterior. |

### 📖 La Analogía del Contenedor

Imagina que trabajas en el archivo del Congreso. Cada variable es una **caja** con una etiqueta:
- La **etiqueta** es el nombre de la variable (ej. `partido_politico`).
- El **contenido** de la caja es el valor (ej. `"Liberal"`).
- Puedes **cambiar** lo que hay dentro sin cambiar la etiqueta (sobrescritura).
- Puedes **leer** el contenido sin destruirlo (con `print()`).

La instrucción de asignación siempre fluye de **derecha a izquierda**: primero Python evalúa lo que hay a la derecha del `=`, y luego lo guarda en la variable de la izquierda.

````python
# Crear una variable: asignar un valor a un nombre
partido_politico = "Liberal"  # La caja 'partido_politico' ahora contiene "Liberal"
print(partido_politico)        # Leemos el contenido de la caja

# IMPORTANTE: sin comillas, print muestra el CONTENIDO de la variable
# Con comillas, print muestra el TEXTO LITERAL
print("partido_politico")      # Imprime el texto "partido_politico", no su valor
````

````python
# Sobrescritura: el nuevo valor reemplaza al anterior
departamento = "Cundinamarca"
print("Antes:", departamento)

departamento = "Antioquia"      # El viejo valor se pierde
print("Después:", departamento)
````

````python
# Python es Case Sensitive: estas son TRES variables diferentes
votos = 100
Votos = 200
VOTOS = 300

print(votos)   # 100
print(Votos)   # 200
print(VOTOS)   # 300 — cada una es su propia 'caja'
````

````python
# === Reglas de nomenclatura ===

# ✅ PERMITIDO:
nombre_candidato = "Gustavo"     # snake_case (recomendado en Python)
_voto_secreto = "confidencial"   # puede empezar con guion bajo
edad2 = 45                        # puede contener números (pero no al inicio)

# ❌ PROHIBIDO (descomenta UNA línea a la vez para ver el error):
# 2edad = 45              # No puede empezar con número
# nombre candidato = "A"  # No puede tener espacios
# nombre-candidato = "A"  # No puede tener guiones medios
````

````python
# Estilos de escritura (convenciones de la comunidad)
mi_variable_nueva = 10      # snake_case → El RECOMENDADO en Python
miVariableNueva = 10        # camelCase  → Común en JavaScript
MiVariableNueva = 10        # PascalCase → Para nombres de clases

# En este curso usaremos SIEMPRE snake_case
print("Convención elegida: snake_case")
````

# Tema 03 — Asignación Múltiple de Variables

### 🎯 Objetivo
Aprender atajos para crear varias variables a la vez y descubrir cómo el operador `+` cambia de comportamiento según el tipo de dato. Este tema solo usa lo que ya conoces: `print()` y variables.

### 🧠 Conceptos clave

| Concepto | Definición |
|---|---|
| **Asignación múltiple** | Definir varias variables con valores en una sola línea. |
| **Concatenación** | Unir dos o más textos (strings) en uno solo con `+`. |
| **Tipo de dato** | Clasificación que determina qué operaciones puede realizar una variable (texto vs. número). |

````python
# Asignación múltiple: valores distintos (posicional)
# Cada valor se asigna a su variable correspondiente por posición
departamento, municipio, region = "Antioquia", "Medellín", "Andina"

print(departamento)  # Antioquia
print(municipio)     # Medellín
print(region)        # Andina
````

````python
# Asignación múltiple: mismo valor para todas las variables
estado_proyecto = resultado_votacion = clasificacion = "Pendiente"

print(estado_proyecto)       # Pendiente
print(resultado_votacion)    # Pendiente
print(clasificacion)         # Pendiente
````

````python
# === Aprendizaje Comparativo: print con COMA vs print con + ===

candidato = "María"
cargo = "Alcaldesa"

# COMA → Python agrega un espacio automático entre los elementos
print(candidato, "fue elegida", cargo)
# Resultado: María fue elegida Alcaldesa

# + → Concatena (pega) todo SIN espacios. Debes agregarlos tú
print(candidato + " fue elegida " + cargo)
# Resultado: María fue elegida Alcaldesa
````

````python
# === El + cambia de comportamiento según el TIPO DE DATO ===

# Con TEXTO (strings): el + CONCATENA (une los textos)
codigo_dane = "05" + "001"  # Código DANE de Medellín
print("Código DANE:", codigo_dane)  # "05001" (es texto, no un número)

# Con NÚMEROS: el + SUMA (hace aritmética)
votos_zona_1 = 1500
votos_zona_2 = 2300
total_votos = votos_zona_1 + votos_zona_2
print("Total de votos:", total_votos)  # 3800 (no "15002300")
````

### 📝 TL;DR
- Puedes asignar valores a varias variables en una sola línea usando comas.
- `print(x, y)` agrega espacios automáticamente; `print(x + y)` no.
- El `+` con textos **concatena** (une); con números **suma** (aritmética).
- `"5" + "6"` da `"56"` (texto); `5 + 6` da `11` (número). ¡El tipo de dato importa!

# Tema 04 — Tipos de Datos: Strings, Números y Colecciones

### 🧠 Conceptos clave

| Concepto | Definición |
|---|---|
| **String (`str`)** | Texto. Se escribe entre comillas. |
| **Integer (`int`)** | Número entero (sin decimales). |
| **Float (`float`)** | Número con decimales. |
| **Booleano (`bool`)** | Valor lógico: solo `True` o `False`. |
| **Lista** | Colección ordenada y **modificable**. Se escribe con `[]`. |
| **Tupla** | Colección ordenada e **inmodificable**. Se escribe con `()`. |
| **Diccionario** | Colección de pares clave-valor. Se escribe con `{}`. |
| **Set (Conjunto)** | Colección **sin duplicados** ni orden fijo. Se escribe con `{}`. |
| **Mutable** | Que se puede modificar después de creado (listas, diccionarios, sets). |
| **Inmutable** | Que NO se puede modificar una vez creado (tuplas, strings). |

### 📖 Explicación: El censo de datos de Python

Piensa en los datos que maneja el DANE en un censo: nombres de personas (texto), edades (números enteros), ingresos promedio (números decimales), si votó o no (verdadero/falso), lista de municipios visitados (colección). Python tiene un tipo de dato para **cada uno** de estos escenarios. La función `type()` te permite preguntarle a Python "¿qué tipo de dato es esto?".

````python
# === STRINGS (Texto) ===
# Tres formas de definir texto en Python
nombre = 'Juan Pablo'           # Comillas simples
apellido = "Restrepo"           # Comillas dobles (equivalente)
discurso = """Este es un discurso
que ocupa varias líneas,
como un comunicado oficial."""   # Comillas triples: texto multilínea

print(nombre)
print(apellido)
print(discurso)
````

````python
# === NÚMEROS ===
poblacion_medellin = 2569007     # int → entero (sin decimales)
tasa_desempleo = 12.5            # float → decimal (con punto)
numero_complejo = 3 + 2j         # complex → para matemáticas avanzadas

# type() te dice qué tipo de dato es una variable
print(type(poblacion_medellin))  # <class 'int'>
print(type(tasa_desempleo))      # <class 'float'>
print(type(numero_complejo))     # <class 'complex'>
````

````python
# === BOOLEANOS (Lógicos) ===
# Solo dos valores posibles, siempre con mayúscula inicial
proyecto_aprobado = True
tiene_quorum = False

print(type(proyecto_aprobado))  # <class 'bool'>
print(proyecto_aprobado)         # True
````

````python
# === COLECCIONES ===

# LISTA [] 
# Como una lista de asistencia: puedes tachar y agregar nombres
partidos = ["Liberal", "Conservador", "Verde", "Polo"]
partidos.append("Pacto Histórico")  # Agregar un elemento
print("Lista:", partidos)

print("---")

# TUPLA ()
# Como los resultados oficiales: una vez publicados, no cambian
resultados_oficiales = ("Candidato A", 8500000, "Candidato B", 6200000)
print("Tupla:", resultados_oficiales)
# resultados_oficiales.append("Candidato C")  # ❌ Esto daría error
````

````python
# DICCIONARIO {} → Pares clave: valor (como una ficha de un candidato)
candidato = {
    "nombre": "Ana María",
    "partido": "Verde",
    "votos": 150000,
    "electo": True
}
print("Nombre:", candidato["nombre"])
print("Votos:", candidato["votos"])

print("---")

# SET {} → Sin duplicados y sin orden fijo
# Si repites un elemento, Python lo descarta automáticamente
municipios_visitados = {"Medellín", "Bello", "Envigado", "Medellín", "Bello"}
print("Municipios únicos:", municipios_visitados)  # Solo 3 elementos
````

### 📝 TL;DR
- **Strings**: texto entre comillas. Comillas triples para multilínea.

# Tema 05 — Manipulación y Conversión de Tipos de Datos (Casting)

### 🎯 Objetivo
Aprender a transformar datos de un tipo a otro (casting) y entender los riesgos de esa transformación, especialmente la pérdida de información al convertir de `float` a `int`.

### 🧠 Conceptos clave

| Concepto | Definición |
|---|---|
| **Casting** | Convertir una variable de un tipo a otro (ej. `float` → `int`). |
| **Truncar** | Eliminar la parte decimal **sin redondear**. `int(3.9)` → `3`, no `4`. |
| **`type()`** | Función que revela el tipo de dato de una variable. |
| **Módulo** | Archivo con funciones extra que puedes importar (ej. `random`). |

### 📖 Explicación

En estadística descriptiva ya trabajaste con datos cuantitativos continuos y discretos. El **casting** en Python es similar a decidir si un dato se reporta con decimales o sin ellos. Pero cuidado: si la tasa de participación electoral es `67.8%` y la conviertes a entero, Python te dará `67`, no `68`. Esa diferencia del 0.8% puede ser significativa en un análisis político.

````python
# Inspección de tipos con type()
participacion = 67.8
total_votantes = 35000000
nombre_eleccion = "Presidenciales 2022"

print(type(participacion))      # <class 'float'>
print(type(total_votantes))     # <class 'int'>
print(type(nombre_eleccion))    # <class 'str'>
````

````python
# === CASTING: Conversiones de tipo ===

# De entero a flotante → Seguro, no se pierde información
votos = 15000
votos_float = float(votos)
print(votos_float, type(votos_float))  # 15000.0 <class 'float'>

print("---")

# ⚠️ De flotante a entero → PELIGRO: truncamiento (NO redondea)
tasa_aprobacion = 67.8
tasa_entera = int(tasa_aprobacion)
print("Original:", tasa_aprobacion)  # 67.8
print("Truncada:", tasa_entera)       # 67 — ¡Se perdió el .8!

# Otro ejemplo dramático:
print("int(99.99) =", int(99.99))     # 99, NO 100
````

````python
# De texto a número (útil cuando lees datos de un archivo)
texto_votos = "42500"   # Esto es texto, no un número
print(type(texto_votos))  # <class 'str'>

numero_votos = int(texto_votos)  # Convertir a entero
print(type(numero_votos))  # <class 'int'>
print(numero_votos + 1000) # Ahora sí podemos sumar: 43500
````

````python
# Generación de números aleatorios (útil para simulaciones)
import random  # 'import' carga funciones adicionales

# Simular el resultado de una encuesta con margen de error
intencion_voto_base = 45  # 45% base
margen_error = random.randrange(-5, 6)  # Número aleatorio entre -5 y +5
resultado_simulado = intencion_voto_base + margen_error

print(f"Intención de voto simulada: {resultado_simulado}%")
print(f"(Margen aplicado: {margen_error} puntos)")

# Ejecuta esta celda varias veces: ¡el resultado cambia cada vez!
````

````python
# Textos multilínea con comillas triples
comunicado = """COMUNICADO DE PRENSA
Fecha: 15 de marzo de 2025
La Registraduría Nacional informa que
el proceso electoral se desarrolló con normalidad."""

print(comunicado)
print("---")
print("Caracteres totales:", len(comunicado))  # len() cuenta TODO
````

````python
# Buscar subcadenas con 'in' y 'not in'
titulo_proyecto = "Reforma Tributaria Estructural 2025"

print("¿Contiene 'Tributaria'?", "Tributaria" in titulo_proyecto)     # True
print("¿Contiene 'tributaria'?", "tributaria" in titulo_proyecto)     # False ← ¡Case Sensitive!
print("¿NO contiene 'Pensional'?", "Pensional" not in titulo_proyecto) # True
````

````python
# Transformación y limpieza de datos
entrada_usuario = "  medellín  "  # Dato con espacios accidentales

print("Original:   >", entrada_usuario, "<")
print("Mayúsculas:", entrada_usuario.upper())
print("Minúsculas:", entrada_usuario.lower())
print("Limpio:     >", entrada_usuario.strip(), "<")  # Sin espacios extra

# Buena práctica: SIEMPRE limpiar antes de guardar
dato_limpio = entrada_usuario.strip().lower()
print("Dato final:", dato_limpio)
````

# Tema 07 — Slicing, Replace y Split

### 🧠 Conceptos clave

| Concepto | Definición |
|---|---|
| **Índice** | Posición de un carácter dentro del texto. Python empieza desde **0**, no desde 1. |
| **Slicing** | Extraer una porción con `[inicio:fin]`. El `fin` es **exclusivo** (no se incluye). |
| **`replace()`** | Sustituye todas las coincidencias de un texto por otro. |
| **`split()`** | Divide un string en una lista usando un carácter separador. |
| **Normalización** | Convertir texto a un formato estándar (ej. todo en minúsculas) para comparar. |

````python
# === ÍNDICES: cada carácter tiene una posición ===
# Posiciones: 0 1 2 3 4 5 6 7
municipio = "Medellín"

print(municipio[0])      # 'M' — primer carácter (posición 0)
print(municipio[1])      # 'e' — segundo carácter (posición 1)
print(municipio[-1])     # 'n' — último carácter (desde el final)
````

````python
# === SLICING: Extraer porciones de texto ===
municipio = "Medellín"

# [inicio:fin] → el fin NO se incluye
print(municipio[0:4])    # 'Mede' — posiciones 0, 1, 2, 3 (la 4 se excluye)

# Atajos:
print(municipio[:4])     # 'Mede' — desde el inicio hasta la posición 3
print(municipio[4:])     # 'llín' — desde la posición 4 hasta el final

# Índices negativos: contar desde el final
print(municipio[-4:])    # 'llín' — los últimos 4 caracteres
````

````python
# === REPLACE: Sustituir texto ===
titulo = "Análisis político del departamento de Caldas"

nuevo_titulo = titulo.replace("Caldas", "Antioquia")
print("Original:", titulo)
print("Modificado:", nuevo_titulo)
````

# Tema 08 — Booleanos en Profundidad: Veracidad, Casting y Validación

### 🧠 Conceptos clave

|---|---|
| **Casting a `bool()`** | Evalúa si un objeto se considera "verdadero" o "falso" según las reglas de Python. |
| **`None`** | Valor especial que representa la ausencia de dato. Evalúa como `False`. |
| **`isinstance()`** | Verifica si una variable pertenece a un tipo de dato específico. |

### 📖 La Regla de Oro

Piénsalo como una urna electoral: si la urna tiene **algo adentro** (votos, papeles, lo que sea), se considera `True`. Si la urna está **vacía**, es `False`. El número **0** es como una casilla sin marcar: también `False`.

````python
# Las comparaciones generan booleanos automáticamente
votos_a_favor = 55
votos_necesarios = 50

aprobado = votos_a_favor > votos_necesarios  # Esto produce True o False
print("¿Proyecto aprobado?", aprobado)         # True
print("Tipo de dato:", type(aprobado))          # <class 'bool'>
````

````python
# === REGLA DE ORO: ¿Qué es True y qué es False? ===

# Cualquier cosa con CONTENIDO → True
print("String con texto:", bool("Democracia"))   # True
print("Número distinto de 0:", bool(42))          # True
print("Lista con elementos:", bool([1, 2, 3]))    # True

print("---")

# Lo VACÍO o NULO → False
print("String vacío:", bool(""))     # False
print("El número 0:", bool(0))       # False
print("Lista vacía:", bool([]))      # False
print("None:", bool(None))           # False
````

````python
# isinstance(): verificar tipos antes de operar
# Evita errores cuando no sabes qué tipo de dato recibes
votos = 35000
porcentaje = 67.5
partido = "Liberal"

print("¿votos es int?", isinstance(votos, int))          # True
print("¿porcentaje es float?", isinstance(porcentaje, float))  # True
print("¿partido es str?", isinstance(partido, str))       # True
print("¿votos es str?", isinstance(votos, str))           # False
````

# Tema 09 — Operadores Matemáticos y Precedencia

### 🧠 Operadores aritméticos

| Operador | Función | Ejemplo | Resultado |
|---|---|---|---|
| `+` | Suma | `5 + 3` | `8` |
| `-` | Resta | `5 - 3` | `2` |
| `*` | Multiplicación | `5 * 3` | `15` |
| `/` | División (siempre float) | `10 / 3` | `3.333...` |
| `%` | Módulo (residuo) | `10 % 3` | `1` |
| `**` | Potencia | `2 ** 3` | `8` |
| `//` | División entera | `10 // 3` | `3` |

### Precedencia (orden de resolución)
1. Paréntesis `()`
2. Potencias `**`
3. Multiplicación, divisiones y módulo `*`, `/`, `//`, `%`
4. Suma y resta `+`, `-`

````python
# Cálculo con datos electorales
votos_candidato_a = 8500000
votos_candidato_b = 6200000
votos_totales = votos_candidato_a + votos_candidato_b

# Porcentaje de votación (nótese los paréntesis para controlar el orden)
porcentaje_a = (votos_candidato_a / votos_totales) * 100
porcentaje_b = (votos_candidato_b / votos_totales) * 100

print(f"Candidato A: {porcentaje_a:.1f}%")  # :.1f → un decimal
print(f"Candidato B: {porcentaje_b:.1f}%")
````

````python
# === Aprendizaje Comparativo: División estándar vs entera ===
curules_totales = 108
partidos = 5

print("División estándar /:", curules_totales / partidos)   # 21.6 (float)
print("División entera //:", curules_totales // partidos)   # 21 (sin decimales)
print("Residuo (módulo) %:", curules_totales % partidos)    # 3 curules sobrantes
````

````python
# === MÓDULO: ¿Par o impar? ===
# Si un número dividido entre 2 tiene residuo 0, es par
numero_concejales = 15
residuo = numero_concejales % 2  # ¿Cuánto sobra al dividir entre 2?

print(f"{numero_concejales} concejales")
print(f"Residuo al dividir entre 2: {residuo}")
print(f"¿Es impar (permite desempate)? {residuo != 0}")  # True
````

````python
# === PRECEDENCIA: ¡Los paréntesis cambian todo! ===
resultado_sin = 2 + 3 * 4     # Python resuelve: 3*4=12, luego 2+12=14
resultado_con = (2 + 3) * 4   # Python resuelve: 2+3=5, luego 5*4=20

print("Sin paréntesis: 2 + 3 * 4 =", resultado_sin)  # 14
print("Con paréntesis: (2 + 3) * 4 =", resultado_con)  # 20
print("Misma regla que en matemáticas: multiplicación antes que suma")
````

# Tema 10 — Operadores de Comparación y Lógicos

### 🧠 Operadores de comparación

| Operador | Significado | Ejemplo | Resultado |
|---|---|---|---|
| `==` | Igual a | `5 == 5` | `True` |
| `!=` | Diferente de | `5 != 3` | `True` |
| `>` | Mayor que | `5 > 3` | `True` |
| `<` | Menor que | `5 < 3` | `False` |
| `>=` | Mayor o igual | `5 >= 5` | `True` |
| `<=` | Menor o igual | `5 <= 3` | `False` |

### Operadores lógicos

| Operador | Función | Regla |
|---|---|---|
| `and` | Conjunción | **Todas** deben ser `True` para dar `True`. |
| `or` | Disyunción | Basta con **una** `True` para dar `True`. |
| `not` | Negación | Invierte: `True` → `False`, `False` → `True`. |

````python
# === Comparaciones básicas ===
votos_obtenidos = 55000
umbral = 50000

print("¿Iguales?", votos_obtenidos == umbral)    # False
print("¿Diferentes?", votos_obtenidos != umbral)  # True
print("¿Mayor?", votos_obtenidos > umbral)        # True
print("¿Mayor o igual?", votos_obtenidos >= umbral)  # True

# ⚠️ ERROR COMÚN: confundir = (asignación) con == (comparación)
# x = 5   → guarda 5 en x
# x == 5  → pregunta: ¿x es igual a 5?
````

````python
# === and: TODAS las condiciones deben cumplirse ===
edad = 22
es_colombiano = True
tiene_cedula = True

puede_votar = edad >= 18 and es_colombiano and tiene_cedula
print(f"Edad: {edad}, Colombiano: {es_colombiano}, Cédula: {tiene_cedula}")
print(f"¿Puede votar? {puede_votar}")  # True (todas se cumplen)
````

````python
# === or: basta con que UNA se cumpla ===
edad = 22
es_mayor_70 = edad > 70
vive_en_exterior = False

tiene_excusa = es_mayor_70 or vive_en_exterior
print(f"¿Tiene excusa para no votar? {tiene_excusa}")  # False (ninguna se cumple)

# Cambiemos una condición:
vive_en_exterior = True
tiene_excusa = es_mayor_70 or vive_en_exterior
print(f"¿Y si vive en el exterior? {tiene_excusa}")    # True (una basta)
````

````python
# === not: invierte el resultado ===
esta_inhabilitado = False

puede_ser_candidato = not esta_inhabilitado
print(f"¿Inhabilitado? {esta_inhabilitado}")
print(f"¿Puede ser candidato? {puede_ser_candidato}")  # True

# 'not' es útil cuando la condición NEGATIVA es más fácil de expresar
````

### 📝 TL;DR
- No confundas `=` (asignación) con `==` (comparación).

# Tema 11 — Condicionales: if, elif, else y la Indentación como Ley

### 🎯 Objetivo
Aprender a hacer que tu programa **tome decisiones**. Hasta ahora, tu código se ejecutaba de arriba hacia abajo sin detenerse. Con los condicionales, ciertas instrucciones se ejecutarán **solo si** una condición se cumple. Aquí también entenderás por qué la **indentación** (los espacios al inicio de una línea) es obligatoria en Python.

### 🧠 Conceptos clave

| Concepto | Definición |
|---|---|
| **`if`** | Ejecuta un bloque solo si la condición es `True`. |
| **`elif`** | (Else if) Verifica otra condición si la anterior fue `False`. |
| **`else`** | Se ejecuta cuando **ninguna** condición anterior se cumplió. |
| **Indentación** | Los 4 espacios al inicio de una línea. Definen qué instrucciones **pertenecen** a qué bloque. |
| **Bloque de código** | Grupo de instrucciones que dependen de una misma condición. |
| **`pass`** | Marcador vacío para bloques que aún no tienen lógica (evita errores). |

### 📖 Explicación: La indentación como estructura jerárquica

Piensa en la Constitución colombiana: tiene Títulos, dentro de los Títulos hay Capítulos, dentro de los Capítulos hay Artículos. Cada nivel está "dentro" del anterior. En Python, la indentación cumple la misma función: las instrucciones indentadas (con 4 espacios) **pertenecen al bloque** de la línea que termina en dos puntos (`:`).

```
TÍTULO I (sin indentación)
    Capítulo 1 (4 espacios)
        Artículo 1 (8 espacios)
        Artículo 2 (8 espacios)
    Capítulo 2 (4 espacios)
```

Python evalúa las condiciones de arriba hacia abajo. **En cuanto encuentra una verdadera, ejecuta ese bloque y salta el resto.**

````python
# === Estructura if / elif / else ===
porcentaje_participacion = 62.5

if porcentaje_participacion >= 70:
    print("Participación ALTA")
    print("La legitimidad electoral es fuerte")
elif porcentaje_participacion >= 50:
    print("Participación MEDIA")
    print("Nivel aceptable pero mejorable")
elif porcentaje_participacion >= 30:
    print("Participación BAJA")
    print("Se requiere análisis de causas")
else:
    print("Participación MUY BAJA")
    print("Crisis de legitimidad")

# Solo se ejecutó el bloque del elif >= 50 (porque 62.5 >= 50 es True)
# Los demás bloques fueron ignorados
````

# Tema 12 — Sentencia match (Python 3.10+)

### 🧠 Conceptos clave

| Concepto | Definición |
|---|---|
| **`match`** | Evalúa una expresión y busca coincidencia entre patrones definidos. |
| **`case`** | Define un patrón específico a comparar. |
| **`case _`** | Comodín (wildcard): captura cualquier valor no contemplado. Equivale al `else`. |

### 📖 ¿Cuándo usar `match` en vez de `if-elif`?

Cuando tienes una variable que puede tomar muchos valores específicos (tipos de elección, días de la semana, códigos de departamentos). Con `if-elif` necesitarías repetir la variable en cada línea; con `match` la escribes una sola vez.

````python
# === match-case: Clasificar tipo de elección ===
tipo_eleccion = "presidencial"

match tipo_eleccion:
    case "presidencial":
        print("Elección de Presidente y Vicepresidente")
        print("Periodicidad: cada 4 años")
    case "legislativa":
        print("Elección de Senado y Cámara")
    case "regional":
        print("Elección de Gobernadores y Alcaldes")
    case "consulta":
        print("Consulta popular o plebiscito")
    case _:
        print("Tipo de elección no reconocido")

# Cambia el valor de tipo_eleccion y ejecuta de nuevo
````

````python
# === match con números: Códigos DANE ===
codigo_departamento = 5

match codigo_departamento:
    case 5:
        print("Antioquia")
    case 11:
        print("Bogotá D.C.")
    case 76:
        print("Valle del Cauca")
    case 8:
        print("Atlántico")
    case _:
        print(f"Código {codigo_departamento}: Consultar tabla DANE")
````

# Tema 13 — Bucles while: Repetir mientras se cumpla una condición

### 🧠 Conceptos clave

| Concepto | Definición |
|---|---|
| **Bucle (`while`)** | Repite un bloque de código mientras una condición sea `True`. |
| **Iteración** | Cada "vuelta" individual del bucle. |
| **Bucle infinito** | Error donde la condición nunca se vuelve `False` (el programa se cuelga). |
| **Incremento** | Actualizar la variable de control para que el bucle avance hacia su fin. |
| **`break`** | Sale del bucle inmediatamente, sin importar la condición. |
| **`continue`** | Salta al inicio de la siguiente vuelta sin ejecutar el código restante. |

### 📖 Explicación

Un bucle `while` es como un **escrutinio**: mientras queden papeletas por contar (`True`), el jurado sigue trabajando. Cuando se terminan (`False`), para. Si olvidas descontar las papeletas contadas (incremento), el jurado contará para siempre — eso es un bucle infinito.

````python
# === Bucle while básico ===
municipios_procesados = 0  # Variable de control: inicia en 0
total_municipios = 5

while municipios_procesados < total_municipios:  # Condición
    municipios_procesados += 1  # ⚠️ Incremento: SIN esto → bucle infinito
    print(f"Procesando municipio {municipios_procesados} de {total_municipios}")

print("¡Todos los municipios han sido procesados!")
````

````python
# === break: Detener el bucle antes de tiempo ===
# Buscar un municipio específico en una lista
municipios = ["Medellín", "Bello", "Envigado", "Itagüí", "Sabaneta"]
indice = 0

while indice < len(municipios):
    if municipios[indice] == "Envigado":
        print(f"¡Encontrado! '{municipios[indice]}' en posición {indice}")
        break  # Ya lo encontramos, dejamos de buscar
    print(f"  Revisando: {municipios[indice]}... no es")
    indice += 1
````

````python
# === continue: Saltar una iteración específica ===
contador = 0

while contador < 10:
    contador += 1
    if contador == 5:
        continue  # Salta el 5: vuelve al while sin ejecutar el print
    print(f"Lote #{contador} procesado")

# Nota: el lote #5 no aparece en la salida
````

````python
# === else: se ejecuta cuando el while termina NATURALMENTE ===
escaños_asignados = 0
escaños_totales = 3

while escaños_asignados < escaños_totales:
    escaños_asignados += 1
    print(f"Escaño {escaños_asignados} asignado")
else:
    # Este bloque se ejecuta solo si el while terminó normalmente
    # (no se ejecuta si el bucle fue interrumpido con break)
    print("Todos los escaños asignados correctamente.")
````

### 📝 TL;DR
- `while` repite código **mientras** la condición sea `True`.
- **Regla de oro**: siempre incluye un incremento para evitar bucles infinitos.
- `break` detiene el bucle por completo.
- `continue` salta al inicio de la siguiente vuelta.
- La posición del incremento dentro del bloque afecta el comportamiento.

# Tema 14 — Bucles for: Recorrido de Secuencias y Listas

### 🧠 Conceptos clave

| Concepto | Definición |
|---|---|
| **`for`** | Recorre cada elemento de una secuencia, uno por uno. |
| **Iterador** | Variable temporal que toma el valor del elemento actual en cada vuelta. |
| **`range()`** | Genera una secuencia numérica: `range(inicio, fin, salto)`. El `fin` no se incluye. |
| **Bucle anidado** | Un `for` dentro de otro: por cada vuelta del externo, el interno se ejecuta completo. |

### 📖 Diferencia con `while`

El `while` repite mientras una condición sea verdadera (no sabes cuántas vueltas dará). El `for` recorre una secuencia definida (sabes exactamente cuántas vueltas dará). Usa `for` cuando tengas una colección que recorrer; usa `while` cuando dependas de una condición dinámica.

**Buena práctica de nomenclatura**: usa **singular** para el iterador y **plural** para la colección: `for municipio in municipios`, `for partido in partidos`.

````python
# === Recorrer una lista ===
departamentos = ["Antioquia", "Cundinamarca", "Valle del Cauca", "Atlántico"]

for departamento in departamentos:  # singular in plural
    print(f"Procesando datos de: {departamento}")
````

````python
# === Recorrer un string (carácter por carácter) ===
sigla = "DANE"

for letra in sigla:
    print(f"Letra: {letra}")
````

````python
# === break y continue en for ===
partidos = ["Liberal", "Conservador", "Verde", "Polo", "Cambio Radical"]

print("--- continue: salta 'Verde' y sigue ---")
for partido in partidos:
    if partido == "Verde":
        continue  # Salta este y pasa al siguiente
    print(f"  {partido}")

print("\n--- break: se detiene al llegar a 'Verde' ---")
for partido in partidos:
    if partido == "Verde":
        break  # Detiene todo el bucle
    print(f"  {partido}")
````

````python
# === range(): Generar secuencias numéricas ===

# Un argumento: range(fin) — empieza en 0
print("Del 0 al 4:")
for i in range(5):
    print(i, end=" ")  # end=" " imprime en la misma línea

print()  # Salto de línea

# Dos argumentos: range(inicio, fin)
print("\nDel 2018 al 2026 (de 4 en 4):")
for año in range(2018, 2027, 4):
    print(f"  Elecciones en {año}")
````

````python
# === Bucles anidados: Cruzar datos de dos colecciones ===
regiones = ["Andina", "Caribe"]
indicadores = ["Pobreza", "Desempleo", "Participación"]

# Por CADA región, se calculan TODOS los indicadores
for region in regiones:
    print(f"Región {region}:")
    for indicador in indicadores:
        print(f"  → Calcular {indicador}")
    print()  # Línea en blanco entre regiones

# Total de líneas: 2 regiones × 3 indicadores = 6 combinaciones
````

````python
# === else en for: confirmar que el recorrido terminó completo ===
municipios_auditar = ["Bello", "Envigado", "Itagüí"]

for municipio in municipios_auditar:
    print(f"Auditando: {municipio}")
else:
    print("Auditoría completada para todos los municipios.")
````

### 📝 TL;DR
- `for` recorre cualquier secuencia: strings, listas, tuplas, rangos.
- `break` termina el bucle; `continue` salta a la siguiente vuelta; `else` confirma la finalización.
- `range(inicio, fin, salto)` genera secuencias numéricas. El fin **no se incluye**.
- Bucles anidados = multiplicación de iteraciones (externo × interno).
- Buena práctica: `for partido in partidos` (singular in plural).
