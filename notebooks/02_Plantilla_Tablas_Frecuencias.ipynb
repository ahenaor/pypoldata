## 📖 Recordatorio teórico

Antes de programar, repasemos los conceptos clave:

| Concepto | Definición | Fórmula |
|----------|-----------|---------|
| **Frecuencia absoluta (fi)** | Número de veces que aparece un valor en los datos | fi = conteo de cada categoría |
| **Frecuencia relativa (hi)** | Proporción que representa cada valor respecto al total | hi = fi / n |
| **Porcentaje (%)** | Frecuencia relativa expresada en porcentaje | % = hi × 100 |

- **Tabla unidimensional:** analiza la distribución de **una sola variable** (ej: ¿cuántos candidatos tiene cada partido?)
- **Tabla bidimensional (contingencia):** analiza la relación entre **dos variables** (ej: ¿cómo se distribuyen los partidos por departamento?)

## Paso 5: Tabla de frecuencia absoluta (fi) — Tabla unidimensional 📋

La **frecuencia absoluta** es simplemente **contar** cuántas veces aparece cada categoría.

**Pregunta politológica:** *¿Cuántos registros de candidato-municipio tiene cada partido en la circunscripción territorial?*

En Pandas, usamos el método `.value_counts()` sobre una columna.

### 💡 ¿Qué nos dice esto?

Cada número representa **cuántos registros candidato-municipio** tiene ese partido. Un partido con más registros tiene presencia en más municipios y/o presenta más candidatos.

> **Nota:** `.value_counts()` ordena automáticamente de mayor a menor frecuencia.

---
## Paso 6: Frecuencia relativa (hi) ➗

La **frecuencia relativa** nos dice **qué proporción del total** representa cada categoría.

**Fórmula:** `hi = fi / n`

Donde:
- `fi` = frecuencia absoluta de la categoría
- `n` = total de observaciones

### 💡 ¿Cómo se lee?

Si un partido tiene frecuencia relativa de **0.089**, significa que el **8.9%** del total de registros candidato-municipio corresponde a ese partido.

> **Verificación:** La suma de todas las frecuencias relativas debe ser **1.0** (o muy cercana, por redondeo).

---
## Paso 7: Porcentaje (%) 💯

El porcentaje es simplemente la frecuencia relativa multiplicada por 100. Es más intuitivo para la interpretación.

**Fórmula:** `% = hi × 100`

## Paso 11: Tabla bidimensional (de contingencia) — Dos variables 📊📊

Hasta ahora hemos analizado **una sola variable** a la vez. Pero en ciencia política muchas veces queremos cruzar dos variables para ver su relación.

**Pregunta politológica:** *¿Cómo se distribuyen los partidos principales según el departamento?*

Para esto usamos `pd.crosstab()`, que crea una **tabla de contingencia** (tabla cruzada).

Vamos a trabajar con un ejemplo concreto para que sea más claro: solo los **5 partidos más grandes** y los **5 departamentos con más registros**.

### 💡 ¿Cómo se lee esta tabla?

- **Cada celda** muestra cuántos registros candidato-municipio tiene ese partido en ese departamento.

## Paso 12: Tabla de contingencia con porcentajes 📐

Podemos convertir la tabla de contingencia a porcentajes para facilitar la comparación.

Hay tres formas de calcular porcentajes en una tabla bidimensional:
1. **Porcentaje del total general** (cada celda / total general)
2. **Porcentaje por filas** (cada celda / total de su fila)
3. **Porcentaje por columnas** (cada celda / total de su columna)
