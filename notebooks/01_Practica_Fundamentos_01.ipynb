## PASO 8 — Operadores de comparación: generando booleanos

Los booleanos no solo se crean escribiendo `True` o `False`. La forma más común de obtenerlos es **comparando valores**. Python te devuelve `True` si la comparación se cumple, y `False` si no.

| Operador | Significado | Ejemplo | Resultado |
|---|---|---|---|
| `>` | Mayor que | `10 > 5` | `True` |
| `<` | Menor que | `3 < 2` | `False` |
| `>=` | Mayor o igual que | `5 >= 5` | `True` |
| `<=` | Menor o igual que | `4 <= 3` | `False` |
| `==` | Igual a | `7 == 7` | `True` |
| `!=` | Diferente de | `"sí" != "no"` | `True` |

## PASO 12 — Reporte formateado

Para terminar, vamos a generar un reporte de texto bien presentado. Hasta ahora hemos impreso las cosas de manera sencilla, pero mira este truco:

Puedes usar **f-strings** para insertar variables directamente dentro de un texto. Se escriben así:

```python
nombre = "Antioquia"
poblacion = 6677930

```

Resultado: `El departamento de Antioquia tiene 6677930 habitantes`

Y para números decimales, puedes controlar cuántos decimales mostrar con `:.2f` (2 decimales):

```python
porcentaje = 53.27456
print(f"Participación: {porcentaje:.2f}%")
```

Resultado: `Participación: 53.27%`
