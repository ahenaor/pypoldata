## 1. ¿Qué es una función y por qué debería importarme?

### La analogía: una función es una receta

Imaginen que trabajan en un observatorio electoral y cada semana deben calcular el porcentaje de participación de 32 departamentos. Sin funciones, tendrían que escribir la misma fórmula 32 veces. Si un día cambia la fórmula (por ejemplo, deciden excluir votos nulos), tendrían que corregirla en 32 lugares distintos.

Una **función** es como una receta con nombre: la escriben una vez, y después solo necesitan "llamarla" cada vez que la necesiten. Si la receta cambia, la corrigen en un solo lugar y automáticamente se actualiza en todos los análisis que la usen.

### Definición formal

Una **función** es un bloque de código reutilizable que:

- Tiene un **nombre** descriptivo (para poder llamarla después)
- Puede recibir **datos de entrada** (llamados *parámetros* o *argumentos*)
- Ejecuta una **secuencia de instrucciones** con esos datos
- Puede **devolver un resultado** que usaremos más adelante

### ¿Por qué son tan importantes?

| Sin funciones | Con funciones |
|---|---|
| Copiamos y pegamos código muchas veces | Escribimos la lógica una sola vez |
| Si hay un error, hay que corregirlo en muchos lugares | Corregimos en un solo lugar |
| El código es largo y difícil de leer | El código es corto y se lee casi como español |
| Difícil de compartir con colegas | Fácil de compartir: "usa mi función `calcular_nep()`" |

### La estructura básica

```python
def nombre_de_la_funcion(parametro1, parametro2):
    """Descripción breve de qué hace la función."""
    # Aquí va el código que hace el trabajo
    resultado = parametro1 + parametro2
    return resultado
```

Desglosemos cada parte:

- `def` → palabra clave que le dice a Python: "voy a definir una función"
- `nombre_de_la_funcion` → el nombre que elegimos (en minúsculas, con guiones bajos)
- `(parametro1, parametro2)` → los datos que la función necesita para trabajar
- `:` → indica que lo que sigue (indentado) es el cuerpo de la función
- `"""docstring"""` → documentación que explica qué hace la función
- `return` → la palabra clave que devuelve el resultado al exterior

**Importante:** Definir una función no la ejecuta. Es como escribir una receta en un libro: la receta existe, pero el plato solo se cocina cuando alguien la sigue. Para ejecutar la función, la "llamamos" por su nombre.

## 2. Nuestra primera función: sin parámetros

Empecemos con lo más simple posible: una función que no necesita ningún dato de entrada. Solo ejecuta una acción fija cada vez que la llamamos.

**Contexto:** Imaginemos que al inicio de cada reporte electoral queremos imprimir un encabezado estándar.

````python
# --- PASO 1: Definir la función ---
# Aquí le decimos a Python: "guarda estas instrucciones con el nombre 'encabezado_reporte'"

def encabezado_reporte():
    """Imprime el encabezado estándar para reportes electorales."""
    print("=" * 50)
    print("   OBSERVATORIO ELECTORAL")
    print("   Reporte de Análisis de Resultados")
    print("=" * 50)

# En este punto la función EXISTE pero NO se ha ejecutado todavía.
# Python simplemente la guardó en memoria.
````

````python
# --- PASO 2: Llamar (ejecutar) la función ---
# Ahora sí le decimos a Python: "ejecuta las instrucciones guardadas"

encabezado_reporte()
````

````python
# --- PASO 3: La ventaja de reutilizar ---
# Podemos llamarla cuantas veces queramos.
# Si mañana cambia el nombre del observatorio, solo editamos la función una vez.

encabezado_reporte()
print("\nDepartamento: Antioquia")
print("Elección: Gobernación 2023")
print()

encabezado_reporte()
print("\nDepartamento: Valle del Cauca")
print("Elección: Gobernación 2023")
````

### Reflexión

Noten que los paréntesis vacíos `()` son obligatorios tanto al definir como al llamar la función. Son ellos los que le dicen a Python "esto es una función, ejecútala". Sin los paréntesis, Python simplemente muestra que el objeto existe, pero no lo ejecuta.

Prueben escribir `encabezado_reporte` sin paréntesis en la celda de abajo para ver la diferencia:

````python
# Sin paréntesis: Python nos dice que es una función, pero NO la ejecuta
print(encabezado_reporte)   # <function encabezado_reporte at 0x...>

print("---")

# Con paréntesis: Python la ejecuta
encabezado_reporte()
````

## 3. Funciones con parámetros: recibir datos de entrada

Las funciones se vuelven realmente útiles cuando pueden recibir datos diferentes cada vez que las llamamos. Esos datos se llaman **parámetros** (cuando definimos la función) o **argumentos** (cuando la llamamos).

### Analogía

Piensen en una fotocopiadora: la máquina siempre hace lo mismo (copiar), pero el resultado cambia según el documento que le metamos. El documento es el "parámetro".

### Ejemplo: Un solo parámetro

**Contexto:** Queremos una función que salude a cualquier analista por su nombre.

````python
def saludar_analista(nombre):
    """
    Saluda a un analista por su nombre.
    
    Parámetros:
        nombre (str): El nombre del analista.
    """
    print(f"Hola, {nombre}. Bienvenido/a al equipo de análisis electoral.")

# Llamamos la función con diferentes argumentos
saludar_analista("Carolina")
saludar_analista("Andrés")
saludar_analista("Valentina")
````

**¿Qué pasó ahí?**

1. Definimos `saludar_analista` con un parámetro llamado `nombre`
2. Cuando escribimos `saludar_analista("Carolina")`, Python reemplaza `nombre` por `"Carolina"` dentro de la función
3. Cada llamada es independiente: `nombre` toma un valor diferente cada vez

### Ejemplo: Varios parámetros

**Contexto:** Queremos describir un municipio con su nombre y número de habitantes.

````python
def describir_municipio(nombre, habitantes, departamento):
    """
    Imprime una descripción breve de un municipio.
    
    Parámetros:
        nombre (str): Nombre del municipio.
        habitantes (int): Número de habitantes.
        departamento (str): Departamento al que pertenece.
    """
    print(f"Municipio: {nombre}")
    print(f"Departamento: {departamento}")
    print(f"Habitantes: {habitantes:,}")
    print()  # línea en blanco para separar

# Probemos con varios municipios
describir_municipio("Medellín", 2_569_007, "Antioquia")
describir_municipio("Quibdó", 130_825, "Chocó")
describir_municipio("Pasto", 392_589, "Nariño")
````

### Detalle importante: el orden de los argumentos

Cuando llamamos una función, Python asigna los argumentos **en el orden** en que los escribimos. Si los pasamos en desorden, el resultado será incorrecto. Veamos:

````python
# CORRECTO: nombre, habitantes, departamento
print("--- Orden correcto ---")
describir_municipio("Cali", 2_227_642, "Valle del Cauca")

# INCORRECTO: los argumentos están desordenados
print("--- Orden incorrecto (¡ojo!) ---")
describir_municipio(2_227_642, "Valle del Cauca", "Cali")
````

### Solución: argumentos nombrados

Para evitar errores de orden, podemos usar **argumentos nombrados** (*keyword arguments*). Así el orden no importa porque le decimos explícitamente a Python qué valor va en qué parámetro:

````python
# Con argumentos nombrados, el orden no importa
describir_municipio(
    departamento="Valle del Cauca",
    habitantes=2_227_642,
    nombre="Cali"
)
````

## 4. Funciones que devuelven valores: `return`

### La diferencia clave: `print` vs `return`

Este es uno de los conceptos que más confusión genera al principio, así que vamos a dedicarle tiempo.

- **`print()`** muestra un texto en la pantalla. Es como hablar en voz alta: los demás escuchan, pero las palabras se desvanecen. No podemos "guardar" lo que se imprimió para usarlo después.

- **`return`** devuelve un valor al código que llamó la función. Es como entregar un documento escrito: el que lo recibe puede guardarlo, modificarlo, pasárselo a otra persona, o hacer cálculos con él.

**Regla práctica:** Si después necesitan usar el resultado de la función para algo más (otro cálculo, una gráfica, un reporte), usen `return`. Si solo quieren mostrar información en pantalla, usen `print`.

### Ejemplo comparativo

````python
# --- Versión con print: el resultado se "pierde" ---

def porcentaje_con_print(votos_candidato, votos_totales):
    """Calcula e imprime el porcentaje. No devuelve nada."""
    porcentaje = (votos_candidato / votos_totales) * 100
    print(f"El porcentaje es: {porcentaje:.2f}%")

# Intentemos guardar el resultado
resultado = porcentaje_con_print(4500, 10000)
print(f"¿Qué guardó 'resultado'? --> {resultado}")
print(f"Tipo de 'resultado': {type(resultado)}")
# resultado es None, porque la función no devolvió nada con return
````

````python
# --- Versión con return: el resultado se puede reutilizar ---

def porcentaje_con_return(votos_candidato, votos_totales):
    """Calcula y devuelve el porcentaje como número."""
    porcentaje = (votos_candidato / votos_totales) * 100
    return porcentaje

# Ahora SÍ podemos guardar y reutilizar el resultado
resultado = porcentaje_con_return(4500, 10000)
print(f"El porcentaje es: {resultado:.2f}%")
print(f"Tipo de 'resultado': {type(resultado)}")

# Y lo podemos usar en cálculos posteriores
print(f"¿Supera el 40%? --> {resultado > 40}")
print(f"El doble sería: {resultado * 2}%")
````

### Aplicación real: Porcentaje de votos válidos

**Contexto:** En toda elección, no todos los votos emitidos son válidos. Algunos son nulos (marcados incorrectamente) y otros son votos en blanco. Para el análisis electoral, necesitamos calcular qué proporción del total fueron votos válidos.

Además, debemos proteger nuestra función contra datos problemáticos: ¿qué pasa si alguien pasa 0 como total de votos? Tendríamos una división por cero, que hace que Python lance un error.

````python
def ratio_votos(votos_validos, votos_totales):
    """
    Calcula el porcentaje de votos válidos respecto del total emitido.
    
    Parámetros:
        votos_validos (int): Cantidad de votos válidos.
        votos_totales (int): Cantidad total de votos emitidos.
    
    Retorna:
        float: Porcentaje de votos válidos (0.0 a 100.0).
               Retorna 0.0 si votos_totales es 0 o negativo.
    """
    # Protección contra división por cero o datos inválidos
    if votos_totales <= 0:
        return 0.0
    
    return (votos_validos / votos_totales) * 100

# --- Probemos con casos diferentes ---

# Caso normal
print("Caso normal:")
print(f"  4,500 de 5,000 = {ratio_votos(4500, 5000)}%")

# Caso de mesa sin votos (protección contra error)
print("\nCaso mesa vacía:")
print(f"  3,000 de 0 = {ratio_votos(3000, 0)}%")

# Caso 100%
print("\nCaso todos válidos:")
print(f"  8,000 de 8,000 = {ratio_votos(8000, 8000)}%")
````

## 5. Funciones con lógica condicional: tomar decisiones

Las funciones pueden incluir estructuras `if/elif/else` para tomar decisiones basadas en los datos que reciben. Esto es especialmente útil en análisis político, donde frecuentemente necesitamos **clasificar** datos en categorías.

### Ejemplo: Clasificar un resultado electoral

````python
def clasificar_resultado(porcentaje):
    """
    Clasifica un resultado electoral según el porcentaje obtenido.
    
    Parámetros:
        porcentaje (float): Porcentaje de votos obtenido (0-100).
    
    Retorna:
        str: Categoría del resultado.
    """
    if porcentaje >= 50:
        return "Mayoría absoluta"
    elif porcentaje >= 30:
        return "Buena base electoral"
    elif porcentaje >= 15:
        return "Fuerza significativa"
    else:
        return "Fuerza minoritaria"

# --- Probemos con diferentes escenarios ---
# Simulamos candidatos con distintos porcentajes

candidatos = [
    ("Candidata A", 52.3),
    ("Candidato B", 31.7),
    ("Candidata C", 22.1),
    ("Candidato D", 8.4),
]

for nombre, pct in candidatos:
    categoria = clasificar_resultado(pct)
    print(f"{nombre}: {pct}% → {categoria}")
````

### ¿Por qué `return` y no `print` dentro del `if`?

Noten que usamos `return` dentro de cada rama del `if`. Esto es intencional: queremos que la función nos **entregue** la categoría como texto, para poder usarla después (en un gráfico, en una tabla, en otro cálculo). Si usáramos `print`, la categoría solo aparecería en pantalla y no podríamos hacer nada más con ella.

**Dato importante:** Cuando Python encuentra un `return`, la función **se detiene inmediatamente**. No ejecuta ninguna línea más. Por eso no necesitamos `else` en cada caso: si entró al primer `if` y ejecutó su `return`, nunca llega al `elif`.

## 6. Valores por defecto en parámetros

A veces queremos que una función tenga un comportamiento "estándar" que se pueda modificar opcionalmente. Para esto existen los **valores por defecto**: le asignamos un valor al parámetro en la definición, y si quien llama la función no pasa ese argumento, Python usa el valor por defecto.

**Contexto:** Queremos una función que formatee porcentajes. Normalmente queremos 1 decimal, pero a veces necesitamos más precisión.

````python
def formatear_porcentaje(valor, decimales=1):
    """
    Formatea un número como porcentaje con texto.
    
    Parámetros:
        valor (float): El número a formatear.
        decimales (int): Cantidad de decimales. Por defecto, 1.
    
    Retorna:
        str: El porcentaje formateado como texto.
    """
    return f"{valor:.{decimales}f}%"

# Sin especificar decimales: usa el valor por defecto (1)
print(formatear_porcentaje(45.678))       # "45.7%"

# Especificando 2 decimales
print(formatear_porcentaje(45.678, 2))    # "45.68%"

# Especificando 0 decimales
print(formatear_porcentaje(45.678, 0))    # "46%"

# Con argumento nombrado (más legible)
print(formatear_porcentaje(45.678, decimales=3))  # "45.678%"
````

## 7. Composición de funciones: construir análisis en capas

### El concepto clave

El verdadero poder de las funciones aparece cuando **combinamos funciones simples para construir análisis complejos**. Cada función hace una tarea pequeña y bien definida, y luego las conectamos como piezas de un rompecabezas.

Esto se llama **composición de funciones** y es la base de todo buen software, incluido el análisis de datos.

### Analogía

Piensen en una cadena de producción:
1. La primera estación corta la tela → función `ratio_votos()`
2. La segunda estación cose la tela → función `clasificar_resultado()`
3. La tercera estación empaca el producto → función `generar_informe()`

El resultado de cada estación es la entrada de la siguiente.

### Ejemplo práctico: Informe electoral completo

Vamos a usar las funciones que ya creamos (`ratio_votos` y `clasificar_resultado`) dentro de una nueva función más completa.

````python
def generar_informe(municipio, votos_validos, votos_totales):
    """
    Genera un diccionario con el informe electoral de un municipio.
    Combina las funciones ratio_votos() y clasificar_resultado().
    
    Parámetros:
        municipio (str): Nombre del municipio.
        votos_validos (int): Votos válidos emitidos.
        votos_totales (int): Total de votos emitidos.
    
    Retorna:
        dict: Diccionario con toda la información del informe.
    """
    # Paso 1: calcular el porcentaje (usando nuestra función anterior)
    porcentaje = ratio_votos(votos_validos, votos_totales)
    
    # Paso 2: clasificar el resultado (usando otra función anterior)
    categoria = clasificar_resultado(porcentaje)
    
    # Paso 3: empaquetar todo en un diccionario
    informe = {
        "municipio": municipio,
        "votos_validos": votos_validos,
        "votos_totales": votos_totales,
        "porcentaje": round(porcentaje, 2),
        "categoria": categoria
    }
    
    return informe

# --- Generemos informes para varios municipios ---

informe_1 = generar_informe("Medellín", 45000, 80000)
informe_2 = generar_informe("Quibdó", 12000, 25000)
informe_3 = generar_informe("Apartadó", 8500, 15000)

# Mostremos cada informe de manera legible
for informe in [informe_1, informe_2, informe_3]:
    print(f"📍 {informe['municipio']}")
    print(f"   Votos válidos: {informe['votos_validos']:,} de {informe['votos_totales']:,}")
    print(f"   Porcentaje: {informe['porcentaje']}%")
    print(f"   Categoría: {informe['categoria']}")
    print()
````

### ¿Por qué esto es tan poderoso?

Observen lo que logramos:

1. **`ratio_votos()`** solo sabe dividir y proteger contra ceros
2. **`clasificar_resultado()`** solo sabe clasificar un número en categorías
3. **`generar_informe()`** no sabe calcular ni clasificar; solo sabe **coordinar** las otras dos funciones y empaquetar el resultado

Cada función tiene **una sola responsabilidad**. Si mañana cambiamos los rangos de clasificación, solo editamos `clasificar_resultado()` y todos los informes se actualizan automáticamente.

Este principio se llama **responsabilidad única** y es una de las mejores prácticas en programación.

## 8. Procesando múltiples datos con funciones y listas

En la práctica, rara vez analizamos un solo municipio o una sola elección. Normalmente tenemos **listas de datos** que necesitamos procesar de forma sistemática. Aquí es donde las funciones brillan: las combinamos con bucles `for` para aplicar el mismo análisis a muchos casos.

**Contexto:** Tenemos datos de participación electoral de varios departamentos y queremos analizar cada uno.

````python
# Datos de ejemplo: (departamento, votos_validos, votos_totales)
datos_departamentos = [
    ("Antioquia",      3_200_000, 4_500_000),
    ("Cundinamarca",   1_800_000, 2_100_000),
    ("Valle del Cauca", 1_500_000, 2_800_000),
    ("Bolívar",          450_000, 1_200_000),
    ("Chocó",            120_000,   350_000),
]

# Aplicamos nuestra función a TODOS los departamentos
print("RESULTADOS ELECTORALES POR DEPARTAMENTO")
print("=" * 55)

informes = []  # lista vacía para guardar todos los informes

for depto, validos, totales in datos_departamentos:
    informe = generar_informe(depto, validos, totales)
    informes.append(informe)  # guardamos cada informe en la lista
    print(f"{depto:<20} {informe['porcentaje']:>6}%  →  {informe['categoria']}")

print("=" * 55)
````

````python
# Ahora podemos hacer análisis SOBRE los informes generados
# Por ejemplo: ¿cuál departamento tuvo mayor porcentaje?

mejor = max(informes, key=lambda x: x["porcentaje"])
peor = min(informes, key=lambda x: x["porcentaje"])

print(f"Mayor porcentaje de votos válidos: {mejor['municipio']} ({mejor['porcentaje']}%)")
print(f"Menor porcentaje de votos válidos: {peor['municipio']} ({peor['porcentaje']}%)")

# Promedio general
promedio = sum(inf["porcentaje"] for inf in informes) / len(informes)
print(f"Promedio nacional: {promedio:.1f}%")
````

## 9. Buenas prácticas: escribir funciones que otros (y tú del futuro) puedan entender

### 9.1 Docstrings: documentar siempre

Un **docstring** es un texto entre comillas triples (`"""`) justo al inicio de la función. Describe qué hace, qué recibe y qué devuelve. Python puede leerlo con `help()`.

````python
def indice_competitividad(votos_primero, votos_segundo):
    """
    Calcula qué tan reñida fue una elección entre los dos primeros candidatos.
    
    Un valor cercano a 1 significa que los dos candidatos obtuvieron
    cantidades similares de votos (elección muy reñida). Un valor
    cercano a 0 significa que el primero ganó por amplio margen.
    
    Parámetros:
        votos_primero (int): Votos del candidato ganador.
        votos_segundo (int): Votos del segundo candidato.
    
    Retorna:
        float: Índice de competitividad entre 0.0 y 1.0.
    """
    if votos_primero <= 0:
        return 0.0
    return votos_segundo / votos_primero

# Ahora cualquiera puede consultar qué hace esta función
help(indice_competitividad)
````

````python
# Probemos el índice con distintos escenarios

# Elección muy reñida
ic1 = indice_competitividad(50200, 49800)
print(f"Elección reñida:   {ic1:.4f}  (cercano a 1 = muy competitiva)")

# Elección con margen amplio
ic2 = indice_competitividad(70000, 30000)
print(f"Margen amplio:     {ic2:.4f}  (lejano de 1 = poco competitiva)")

# Elección intermedia
ic3 = indice_competitividad(55000, 45000)
print(f"Margen intermedio: {ic3:.4f}")
````

1. Crea una función `calcular_participacion(inscritos, votantes)` que:
   - Reciba el número de personas inscritas y el número de votantes
